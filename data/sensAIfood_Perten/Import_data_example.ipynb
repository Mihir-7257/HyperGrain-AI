Example on how to import the data from 'Wheat_sensAIfood_Perten.csv' into a pandas dataframe and plot the spectra.

```python
## substitute the path to the directory where the data is stored (here assuming local directory)
wheat_data  = pd.read_csv('Wheat_sensAIfood_Perten.csv')

## display the first 5 rows of the dataset
display(wheat_data.head(5))

## print dataset dimensions
print(f'\nDataset dimensions: {wheat_data.shape}')

## Define the target moisture and protein values
y_protein = wheat_data['Protein']
y_moisture = wheat_data['Moisture']

## Define the wavelength range and convert to a numpy array
wavelengths = wheat_data.iloc[:, 8:].columns.astype(float).values

## Define the spectra
spectra = wheat_data.iloc[:, 8:].values 
```

```python
## Create a simple plot of the spectra
plt.figure(figsize=(8,3))
plt.plot(wavelengths, spectra.T)
plt.xlabel('Wavelength (nm)')
plt.ylabel('Reflectance')
plt.title('Wheat Spectra')
plt.show()
```
